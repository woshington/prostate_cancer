```python
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights
import torch
import pandas as pd
import os
from utils.dataset import PandasDataset
from utils.models import EfficientNetApi
from utils.metrics import evaluation, format_metrics
```

```python
ROOT_DIR = '../../..'

data_dir = '../../../../dataset'
images_dir = os.path.join(data_dir, 'tiles')
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
```

```python
df_test = pd.read_csv(f"{ROOT_DIR}/data/val.csv")
test_dataset = PandasDataset(images_dir, df_test, transforms=None)
test_loader = torch.utils.data.DataLoader(test_dataset, batch_size=2)
```

```python

for model_name in ["b0-entropy.pth","b0-entropy-00.pth", "b0-entropy-50.pth", "b0-entropy-100.pth"]:
    load_model = efficientnet_b0(weights=EfficientNet_B0_Weights.DEFAULT)
    model = EfficientNetApi(model=load_model, output_dimensions=5, dropout_rate=0.6)
    model = model.to(device)
    model.load_state_dict(
        torch.load(f"models/{model_name}")
    )
    response = evaluation(model, test_loader, device)
    result = format_metrics(response[0])
    print(f"==================================== {model_name}==============")
    print(result)
```

```python
df_test = pd.read_csv(f"{ROOT_DIR}/data/test.csv")
test_dataset = PandasDataset(images_dir, df_test, transforms=None)
test_loader = torch.utils.data.DataLoader(test_dataset, batch_size=2)

model_name = "b0-entropy.pth"
load_model = efficientnet_b0(weights=EfficientNet_B0_Weights.DEFAULT)
model = EfficientNetApi(model=load_model, output_dimensions=5, dropout_rate=0.6)
model = model.to(device)
model.load_state_dict(
    torch.load(f"models/{model_name}")
)
response = evaluation(model, test_loader, device)
```

```python
result = format_metrics(response[0])
print(response[0])
print(result)
```
